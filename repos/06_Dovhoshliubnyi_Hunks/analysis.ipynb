```python
import sqlite3
import os
from pathlib import Path

DB_PATH = Path("aidev.db")

conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.execute("SELECT filename FROM performance_commit_details")
rows = cur.fetchall()
conn.close()

extensions = {Path(row[0]).suffix or '(no extension)' for row in rows}

print(f"Total files: {len(rows)}")
print(f"Unique extensions ({len(extensions)}):")
for ext in sorted(extensions):
    print(f"  {ext}")
```
